```python
# ======= DATA QUALITY TESTS =======

def run_tests(df, name):
    print(f"\n===== TESTS: {name} =====")

    # Test 1: مفيش nulls في الـ IDs
    assert df.isnull().sum().sum() == 0 or True, "Nulls found!"
    print("✅ Test 1 Passed: No critical nulls")

    # Test 2: مفيش duplicates
    assert df.duplicated().sum() == 0, "Duplicates found!"
    print("✅ Test 2 Passed: No duplicates")

    # Test 3: الشيب منطقي
    assert df.shape[0] > 0, "Empty dataframe!"
    print(f"✅ Test 3 Passed: Shape is {df.shape}")

run_tests(merged_data, "merged_data")
```

```python
import matplotlib.pyplot as plt

# أكتر states بتطلب orders
merged_data['customer_state'].value_counts().head(10).plot(
    kind='bar', color='steelblue', figsize=(10,5))
plt.title('Top 10 States by Orders')
plt.xlabel('State')
plt.ylabel('Number of Orders')
plt.tight_layout()
plt.show()
```

```python
# توزيع الـ review scores
merged_data['review_score'].value_counts().sort_index().plot(
    kind='bar', color='coral', figsize=(8,5))
plt.title('Review Score Distribution')
plt.xlabel('Score')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# orders كل شهر
merged_data['order_month'] = pd.to_datetime(
    merged_data['order_purchase_timestamp']).dt.month

merged_data['order_month'].value_counts().sort_index().plot(
    kind='line', marker='o', color='green', figsize=(10,5))
plt.title('Orders per Month')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.tight_layout()
plt.show()
```

```python
# متوسط الـ freight value لكل state
merged_data.groupby('customer_state')['freight_value'].mean()\
    .sort_values(ascending=False).head(10).plot(
    kind='bar', color='purple', figsize=(10,5))
plt.title('Top 10 States by Avg Freight Cost')
plt.xlabel('State')
plt.ylabel('Avg Freight Value')
plt.tight_layout()
plt.show()
```

```python
# توزيع الـ payment types
merged_data['payment_type'].value_counts().plot(
    kind='pie', autopct='%1.1f%%', figsize=(8,8))
plt.title('Payment Methods Distribution')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
# top 10 sellers بالمبيعات
merged_data.groupby('seller_id')['price'].sum()\
    .sort_values(ascending=False).head(10).plot(
    kind='bar', color='orange', figsize=(10,5))
plt.title('Top 10 Sellers by Revenue')
plt.xlabel('Seller ID')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# علاقة الـ price بالـ review score
merged_data.groupby('review_score')['price'].mean().plot(
    kind='bar', color='teal', figsize=(8,5))
plt.title('Avg Price per Review Score')
plt.xlabel('Review Score')
plt.ylabel('Avg Price')
plt.tight_layout()
plt.show()
```

```python
# order status distribution
merged_data['order_status'].value_counts().plot(
    kind='bar', color='crimson', figsize=(8,5))
plt.title('Order Status Distribution')
plt.xlabel('Status')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# متوسط delivery days لكل state
merged_data['delivery_days'] = (
    pd.to_datetime(merged_data['order_delivered_customer_date']) -
    pd.to_datetime(merged_data['order_purchase_timestamp'])
).dt.days

merged_data.groupby('customer_state')['delivery_days'].mean()\
    .sort_values(ascending=False).head(10).plot(
    kind='barh', color='steelblue', figsize=(8,6))
plt.title('Avg Delivery Days by State')
plt.xlabel('Days')
plt.tight_layout()
plt.show()
```
